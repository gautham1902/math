```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.arange(1,6)

suma = np.sum(a)
suma2 = np.sum(a**2)

result1 = suma / suma2
result2 = 1/ suma
print("Option 1 = %s\nOption 2 = %s" %(result1,result2))
```

```python
proda = np.prod(a)
proda2 = np.prod(a**2)

result1 = proda / proda2
result2 = 1/ proda
print("Option 1 = %s\nOption 2 = %s" %(result1,result2))
```

```python
x = np.linspace(-2,2,101)

f = x**2
df = np.diff(f)
dx = np.diff(x)
plt.plot(x,f,label = 'f')
plt.plot(x[:-1],df/dx,label = 'df/dx')
plt.axis([-2,2,-1,2])
plt.legend()
plt.show()
print(df)
print(dx)
```
