```python
import numpy as np
import sympy as sym
from IPython.display import display,Math
```

```python
coeffs = [3,2,-1]

roots = np.roots(coeffs)
print(roots)
```

```python
x = sym.symbols('x')

po = 3*x**2 + 2*x -1

for i in roots:
    display(Math('\\text{At x} = %g, \\quad%s  = %g' %(i,sym.latex(po),po.subs(x,i))))
```
