```python
import math
import sympy as sym
from IPython.display import display, Math 
```

```python
x = sym.symbols('x')
expr = (x-1)*(x+3) > 0
```

```python
exp = 3*x/2 + (4-5*x)/3 < 2 - 5*(2-x)/4
display(sym.solve(exp,x))
```

```python
from sympy.abc import x

p1 = 2*x**3 + x**2 - x
p2 = x**3 - x**4 - 4*x**2

display(Math('(%s) - (%s) = %s' %(sym.latex(p1),sym.latex(p2),sym.latex(p1-p2))))
```

```python
p1 = sym.Poly(2*x**3 + x**2 - x)
print(p1)
```

```python
p1.eval(1) # this is for substuition
p1.coeffs() # this will provide coeff information
```

```python
polyLst = [ sym.Poly(x**2 + 2*x), sym.Poly(-x**3 + 4*x), sym.Poly(x**5 - x**4 + (x/4) + 4) ]

for poli in polyLst:
    if poli.degree()%2 == 0:
        print('The degree of %s is even, and the coefficients sum to %g' %(poli.as_expr(),sum(poli.coeffs())))
    else:
        print('The degree of %s is odd, and there are %s coefficients' %(poli.as_expr(),len(poli.coeffs())) )
```
