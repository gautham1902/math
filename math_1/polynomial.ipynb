```python
from sympy.abc import x
polyLst = [ sym.Poly(x**2 + 2*x), sym.Poly(-x**3 + 4*x), sym.Poly(x**5 - x**4 + (x/4) + 4) ]

for poli in polyLst:
    if poli.degree()%2 == 0:
        print('The degree of %s is even, and the coefficients sum to %g' %(poli.as_expr(),sum(poli.coeffs())))
    else:
        print('The degree of %s is odd, and there are %s coefficients' %(poli.as_expr(),len(poli.coeffs())) )
```

```python
from sympy.abc import x,y
Fxy = 4*x**4 - 3*x**2 + x*y**2 -9*y**3
Gxy = -x**3 + 6*x**2*y +.8*y**3

display(Math(sym.latex(sym.expand(Fxy*Gxy))))
exp = (Fxy*Gxy).subs([(x,5) ,(y,-2)])


exp2 = Fxy.subs([(x,5),(y,-2)])
exp3 = Gxy.subs([(x,5),(y,-2)])

mulRes = exp2 * exp3
print('Polynomial multiply value is %s and Polynomial substitue value is %s' %(exp,mulRes))
```

```python
p1 = 4*x**5 - x
p2 = 2*x**3 - x

display(Math('\\frac{%s}{%s} = %s' %(sym.latex(p1),sym.latex(p2),sym.latex(p1/p2))))
display(Math('\\frac{%s}{%s} = %s' %(sym.latex(p1),sym.latex(p2),sym.latex(sym.expand(p1/p2)))))
display(Math('\\frac{%s}{%s} = %s' %(sym.latex(p1),sym.latex(p2),sym.latex(sym.simplify(p1/p2)))))
```

```python
pp1 = x**6 + 2*x**4 + 6*x -y
pp2 = x**3 + 3

nums = range(5,16)

for i in nums:
    tempNum = pp1.subs(y,i)
    display(Math(' %s  = %s ' %(sym.latex(tempNum/pp2),sym.latex(sym.simplify(tempNum/pp2)))))

    if sym.fraction(sym.simplify(tempNum/pp2))[1] == 1:
        rightans = i
print('Answer satisfy our goal is %g' %rightans)
```
