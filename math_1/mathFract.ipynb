```python
import math
import sympy as sym
from IPython.display import display, Math
import numpy as np
from sympy.abc import x,y
```

```python
a,b,c = sym.symbols('a,b,c')
a = 15
b = 6
c = 3
print(sym.gcd(a*c,b*c))
print(c*math.gcd(a,b))
```

```python
n = 10
m = 15

gcdMat = np.zeros([n,m])+99

for rowi in range(0,n):
    for coli in range(0,m):
        gcdMat[rowi,coli] = math.gcd(rowi+1,coli+1)

display(Math(sym.latex(sym.sympify(gcdMat))))
```

```python
e1 = 4*x - 6
e2 = sym.sin(y)
e3 = x**2 - 9

D = dict(eqnX = [e1,e3], eqnY = [e2])

for keyi in D:
    print('Solutions for equation involving ' + keyi[-1] + ':')
    for eqni in D[keyi]:
        leftpart = sym.latex(eqni) + ' = 0'
        middlepart = '\\quad \\Rightarrow \\quad ' + keyi[-1] + ' = '
        rightpart = sym.latex(sym.solve(eqni))
        display(Math('\\quad \\quad ' +leftpart + middlepart +rightpart))
```

```python
number = range (2,51)

for prm in number:
    fact = sym.factorint(prm)
    ks = list(fact.keys())
    if len((fact)) == 1 and fact[ks[0]] == 1:
        print('%g is a prime number ' %prm)
    else:
        print('%g is a composite number with prime factors: %s' %(prm,list(fact.keys())))
```
