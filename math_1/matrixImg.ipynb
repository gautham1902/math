```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = [[1,2],[1,4]]
plt.imshow(A)
plt.show()
```

```python
A = np.zeros((10,14))

for i in range(0,np.shape(A)[0]):
    for j in range(0, np.shape(A)[1]):
        A[i,j] = 3*i - 4*j

#print(A)
plt.imshow(A)

for i in range(0,np.shape(A)[0]):
    for j in range(0, np.shape(A)[1]):
        plt.text(j,i,int(A[i,j]),horizontalalignment = 'center',verticalalignment = 'center')

plt.set_cmap('twilight_shifted')
plt.show()
```

```python
N = 10
B = np.zeros((N,N))

for i in range(0,N):
    for j in range(0, N):
        B[i,j] = (-1)**(i+j) #PEMDAS

#print(B)
plt.imshow(B)
plt.set_cmap('gray')
plt.tick_params(labelleft=False,labelbottom=False)
plt.show()
```

```python
from matplotlib.patches import Polygon
```

```python
y = np.array([[1,1],[2,3],[3,1]])
p = Polygon(y,facecolor = 'm',alpha = .3)

y1 = np.array([[2,1],[3,3],[4,1]])
p1 = Polygon(y1,facecolor = 'b',alpha = .3,edgecolor = 'k')

fig,ax = plt.subplots()

ax.add_patch(p)
ax.add_patch(p1)
ax.set_xlim([0,4])
ax.set_ylim([0,4])
plt.show()
```

```python
x = np.linspace(-2,2,101)

f = -x**2

pnts = np.vstack((x,f)).T
p = Polygon(pnts, facecolor = 'g',alpha = .2)

y = np.array([[-.5,-4],[-.5,-2.5],[.5,-2.5],[.5,-4]])
p1 = Polygon(y, facecolor = 'k')

fig,ax = plt.subplots()
ax.add_patch(p)
ax.add_patch(p1)
plt.plot(x,f,'k')
plt.plot([-2,2],[-4,-4],'k')
plt.axis('off')
plt.savefig('Test.pdf')
plt.show()


        
```
